## Deleting the Resource group and Resource

```python
import os
from dotenv import load_dotenv, find_dotenv
from azure.identity import DefaultAzureCredential
from azure.mgmt.resource import ResourceManagementClient
```

```python
# Read the environment file to access your secret keys
load_dotenv("CVSKEP.env")
```

```python
from azure.identity import AzureCliCredential

credential = AzureCliCredential()
token = credential.get_token("https://management.azure.com/.default")

print("Authentication successful")
```

```python
# 1. Provide your Subscription ID
SUBSCRIPTION_ID = os.environ["SUBSCRIPTION_ID"]
resource_group_name = os.environ["RESOURCE_GROUP_NAME"]
# 2. Authenticate
credential = DefaultAzureCredential()
# 3. Initialize the Resource Management Client
resource_client = ResourceManagementClient(credential, SUBSCRIPTION_ID)
```

```python
resource_group_name
```

```python
# Delete resource group
delete_operation = resource_client.resource_groups.begin_delete(resource_group_name)
delete_operation.wait()
print("Resource group deleted")
```
